# Which genre of music is the most frequently observed on Spotify’s Top Tracks - USA playlist? How does this compare to the Top Tracks - Global playlist?

I plan on using a retrospective observational study to answer this question. I plan on analyzing the entire population of data because I can observe the entirety of the Top Tracks playlist. I collected this data with the help of AI. I took screenshots of the contents of the playlist and I asked AI to list the artist and track in csv format. I then scraped Wikipedia articles for the genre values, which were entered by humans. Finally, I asked AI to describe the genre of each song that was listed. This may produce some biases because the AI summaries are different than the Wikipedia genre summaries. If the AI summaries are incorrect, then the genre count for the playlists will be incorrect because the wrong genres were examined. If I use the wikipedia genres, then I will most likely need to generalize them myself because they are unique values at this point. How I choose to generalize them may impact the count I observe.

### My Analysis

```python
import pandas as pd
import seaborn as sns
```

```python
usa = pd.read_csv("usa.csv")
globe = pd.read_csv("global.csv")
```

```python
usa.info()
```

```python
usa['Wiki Genre'].unique()
```

```python
freq = usa['AI Genre'].value_counts()
```

```python
freq
```

```python
freq_table = {
    "Genre": ["Pop", "Country", "Hip-Hop"],
    "Percent": [16/50*100, 11/50*100, 9/50*100]
}
```

```python
freq_table_usa = pd.DataFrame(freq_table)
```

```python
freq_table_usa.head()
```

```python
sns.barplot(freq_table_usa, x="Percent", y="Genre")
```

```python
import matplotlib.pyplot as plt

sns.set_theme(palette="pastel")

# Create the pie chart
plt.figure(figsize=(18, 18)) # Adjust figure size
plt.pie(freq_table_usa['Percent'], labels=freq_table_usa['Genre'], autopct='%1.1f%%', startangle=90)
plt.title('Distribution of Categories')
plt.axis('equal') # Equal aspect ratio ensures a circular pie chart
plt.show()
```
