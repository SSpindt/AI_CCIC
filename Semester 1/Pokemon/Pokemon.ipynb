# How does the speed of Pikachu compare to other first generation Pokemon?

I plan on conducting an observational study to answer this question. More specifically, a retrospective study where I collect data from the official Pokedex (https://www.pokemon.com/us/pokedex). This is maintained by the owners of Pokemon, so the data should be complete and up-to-date. More specifically, I will scrape the Pokedex for all first generation Pokemon data. This census approach should limit the amount of bias produced in the results because I will be examining the entirety of the first generation Pokemon data. This data should also be an accurate representation of the entire population because it is curated by the creators of Pokemon.

# My Analysis

```python
import pandas as pd
import seaborn as sns
```

```python
pokemon = pd.read_csv("pokemon.csv")
pokemon.columns
```

```python
first = pokemon[pokemon['generation'] == 1]
```

```python
pikachu = pokemon[pokemon['name'] == "Pikachu"]
```

```python
pikachu
```

```python
sns.histplot(data=first, x="speed")
```

```python
first["speed"].mean()
```

```python
sns.kdeplot(data=first, x="speed")
```

```python
def percentile(data, variable, target):
    n = len(data[variable])
    less_than_data = data[data[variable] <= target]
    return len(less_than_data)/n
```

```python
percentile(first, "speed", 90)
```

# My Answer
Pikachu is faster than many other first generation Pokemon. More specifically, Pikachu is in the 79th percentile of speed, with a speed of 90. The mean speed value for the first generation Pokemon is 70, so Pikachu is 20 points higher than the average Pokemon.
