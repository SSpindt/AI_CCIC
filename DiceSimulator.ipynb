```python
import random
import seaborn as sns
```

```python
results = {
"Side": [1, 2, 3, 4, 5, 6],
"Frequency": [0, 0, 0, 0, 0, 0]
}

for i in range(0, 10):
    num = random.randint(1, 6)
    results["Frequency"][num-1] += 1
```

```python
def roll_dice(small_side, large_side):
    num = random.randint(small_side, large_side)
    return num
```

```python
n = int(input())
for i in range(0, n):
    print(roll_dice(1, 12))
```

```python
sns.barplot(data=results, x="Side", y="Frequency")
```
