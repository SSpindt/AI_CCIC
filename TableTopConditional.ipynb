# In a game of Hanabi, given that the first card drawn is green, what is the probability this card is a 2.

The sample space of this simulation is all green cards in Hanabi. This is more narrow than before because we already know that the first card is green --  that is a given.
One trial of this simulation will be the act of drawing a random green card.
The outcome I will be measuring is how often a green 2 is drawn.

```python
import pandas as pd
import seaborn as sns
import random
```

```python
green_cards = [1, 1, 1, 2, 2, 3, 3, 4, 4, 5] # P(A|B) = 2/10
is_two = 0
not_two = 0
```

```python
for i in range(100):
    idx = random.randint(0, len(green_cards)-1)
    if green_cards[idx] == 2:
        is_two += 1
    else:
        not_two += 1
```

```python
print(is_two/1000)
```

```python
results = {
    "Observation": ["Is Two", "Not Two"],
    "Count": [is_two, not_two]
}
results = pd.DataFrame(results)
```

```python
sns.barplot(results, x="Observation", y="Count")
```

The theoretical probability of pulling a two from the green cards in Hanabi is 1/5 or 20%. This is confirmed by the empirical probability I observed when running my simulation 1000 times. I found the empirical probability to be ~0.22, which is very close to the theoretical probability. In the game of Hanabi, this is neither the most common or rarest green card to observe. For example, the theoretical probability of pulling a green 5 is 1/10.
